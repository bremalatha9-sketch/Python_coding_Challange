```python

#Day 47 – Revision + Coding Challenge
students={"Arun":{"age":25,"grades":[40, 45, 45]},"Banu":{"age":25,"grades":[65, 70, 72]},"Ryan":{"age":26,"grades":[68,79,86]}}
classes={"python","sql","web develper","programmer"}
#print(students)

def student_performance(grades):
  average=round((sum(grades)/len(grades)),2)
  if average >=75:
    return "Distinction"
  elif average >45:
    return "pass"
  else:
    return "fail"

def display_students():
  for name,data in students.items():
    average=round(sum(data["grades"])/len(data["grades"]),2)
    performance=student_performance(data["grades"])
    print(f"student name        :{name}")
    print(f"student age         :{data["age"]}")
    print(f"student grades      :{data["grades"]}")
    print(f"student Average     :{average}")
    print(f"student performance :{performance}")

def add_students():
  name=input("Enter the name    :")
  age=int(input("Enter the age  :"))
  grade1=int(input("Enter the grade1:"))
  grade2=int(input("Enter the grade2:"))
  grade3=int(input("Enter the grade3:"))
  students[name]={"age":age,"grades":[grade1,grade2,grade3]}
  print("students added successfully")

def search_students():
  name=input("Enter the searching name :")
  if name in students:
    print(f"student name           :{name}")
    print(f"student age            :{students[name]["age"]}")
    print(f"student grades         :{students[name]["grades"]}")
    print(f"student performance    :{student_performance(students[name]["grades"])}")
  else:
    print ("student not found")

def delete_students():
  name=input("Enter the deleting name  :")
  if name in students:
    del students[name]
    print("student is deleted")
  else:
    print ("student not found")


while True:

  print("student management system")
  print("1. Display students")
  print("2. Add Students    ")
  print("3. search students ")
  print("4. Delete students  ")
  print("5. Exit")
  choice=input("Enter your choice  :")

  if choice=="1":
    display_students()
  elif choice=="2":
    add_students()
  elif choice=="3":
    search_students()
  elif choice=="4":
    delete_students()
  elif choice=="5":
    print("program completed")
    break
  else:
    print("Invalid choice")





```
