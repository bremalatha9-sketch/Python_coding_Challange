```python
#Day 38 – String Methods and Slicing
#Project Title: Feedback Analyzer
feedback=input("Enter the feedback : ")
word_count=len(feedback.split())
print("number of word : ",word_count)
data_count=feedback.lower().count("data")
print("number of data : ",data_count)
print("print the first 5 characters :", feedback[:5])
print("print the last 5 characters  :", feedback[-5:])
print("print the reverse order      :", feedback[::-1])
#splits() divides a string into parts,len() counts the no of characters
```
