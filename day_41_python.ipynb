```python
#Day 41 – Dictionary, Set, Frozenset
#Project Title: Student Grades & Hobbies Tracker
#Create a dictionary with student names as keys and their grades as values.

student_grade={"Vijay":"A","JAY":"B","Prem":"A"}
print("The students with grade  :", student_grade)

#add or update grades for at least 3 students.

name1=input("Enter the student name1   :")
grade1=input("Enter the student grade1 :")
student_grade[name1]=grade1
name2=input("Enter the student name2   :")
grade2=input("Enter the student grade2 :")
student_grade[name2]=grade2
name3=input("Enter the student name3   :")
grade3=input("Enter the student grade3 :")
student_grade[name3]=grade3
print("The students with grade  :", student_grade)






```

```python
#hobbies and allow the user to input at least 3 hobbies.
student_hobbies=set()
hobby1=input("Enter the students hobby1  :")
student_hobbies.add(hobby1)
hobby2=input("Enter the students hobby2  :")
student_hobbies.add(hobby2)
hobby3=input("Enter the students hobby3  :")
student_hobbies.add(hobby3)
print("The student hobbies are  :", student_hobbies)
#hobbies into a frozenset and display it.
fixed_hobbies=frozenset(student_hobbies)
print("The fixed hobbies are   :"  ,fixed_hobbies)
```
